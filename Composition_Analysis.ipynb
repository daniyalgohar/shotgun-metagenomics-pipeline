```python
from pathlib import Path
import pandas as pd
```

```python
# folder containing all sample TSVs
in_dir = Path("/home/dgohar/res/GenomeMatchinfo")
out_file = in_dir / "all_samples_combined.tsv"
```

```python
# read all TSVs, add sample_id from filename, combine
dfs = []

for f in sorted(in_dir.glob("*.tsv")):
    if f.name == out_file.name:
        continue  # skip previous combined file if it exists
    
    sample_id = f.stem
    df = pd.read_csv(f, sep="\t")
    df.insert(0, "sample_id", sample_id)
    dfs.append(df)

combined = pd.concat(dfs, ignore_index=True)
combined.to_csv(out_file, sep="\t", index=False)

print(combined.shape)
combined.head()
```
